```python
import numpy as np
```

```python
lst_arr1 = [1, 2, 3]
arr1 = np.array(lst_arr1)
lst_arr2 = [4, 5, 6]
arr2 = np.array(lst_arr2)
print(arr1)
print(arr2)
```

```python
dot_product = 0
for i in range(len(lst_arr1)): dot_product += lst_arr1[i] * lst_arr2[i]

dot_product
#here we are calculating the dot product of two arrays using a for loop. The dot product is calculated by multiplying corresponding elements of the two arrays and summing the results. In this case, the dot product of arr1 and arr2 is 32 (1*4 + 2*5 + 3*6 = 4 + 10 + 18 = 32).
```

```python
dot_product = np.dot(arr1, arr2)
dot_product
```

```python
lst_large_arr1 = list(range(1, 100000))
lst_large_arr2 = list(range(1, 100000))

large_arr1 = np.array(lst_large_arr1)
large_arr2 = np.array(lst_large_arr2)
```

```python
import time
start_time = time.time()
dot_product = 0
for i in range(len(lst_large_arr1)):
    dot_product += lst_large_arr1[i] * lst_large_arr2[i]
print('Time taken to calculate the dot product for two large arrays using a loop:', time.time() - start_time)
print()

start_time = time.time()
np.dot(large_arr1, large_arr2)
print('Time taken to calculate the dot product for two large arrays using NumPy:', time.time() - start_time)
```

```python
lst_mat1 = [[1, 2], [3, 4], [5, 6]]
mat1 = np.array(lst_mat1)
lst_mat2 = [[7, 8], [9, 10], [11, 12]]
mat2 = np.array(lst_mat2)
print('Matrix 1:\n', mat1)  
print('Matrix 2:\n', mat2)  
mat1.shape, mat2.shape
```

```python
# Make mat2 have shape (2, 3) instead of (3, 2)
result = np.dot(mat1, mat2.T)  # .T transposes the matrix
print("Matrix multiplication:") 
print(result)
# here calculation formula for each element in the resulting matrix is as follows:
# result [[ 23  29  35], [ 53  67  81], [ 83 105 127]]
# its formula is as follows:
# result[0, 0] = mat1[0, 0] * mat2[0, 0] + mat1[0, 1] * mat2[1, 0] = 1*7 + 2*9 = 23
# result[0, 1] = mat1[0, 0] * mat2[0, 1] + mat1[0, 1] * mat2[1, 1] = 1*8 + 2*10 = 29
# result[0, 2] = mat1[0, 0] * mat2[0, 2] + mat1[0, 1] * mat2[1, 2] = 1*9 + 2*11 = 35
# result[1, 0] = mat1[1, 0] * mat2[0, 0] + mat1[1, 1] * mat2[1, 0] = 3*7 + 4*9 = 53
# result[1, 1] = mat1[1, 0] * mat2[0, 1] + mat1[1, 1] * mat2[1, 1] = 3*8 + 4*10 = 67
# result[1, 2] = mat1[1, 0] * mat2[0, 2] + mat1[1, 1] * mat2[1, 2] = 3*9 + 4*11 = 81
# result[2, 0] = mat1[2, 0] * mat2[0, 0] + mat1[2, 1] * mat2[1, 0] = 5*7 + 6*9 = 83
# result[2, 1] = mat1[2, 0] * mat2[0, 1] + mat1[2, 1] * mat2[1, 1] = 5*8 + 6*10 = 94
```

```python
# If you want element-wise multiplication (same shape)
result = mat1 * mat2  # NOT np.dot()

print("Element-wise multiplication:")
print(result)
print("Shape:", result.shape)  # (3, 2)
```

```python
#The @ operator is specifically designed for matrix multiplication. It's the same as np.matmul() and similar to np.dot() for 2D arrays.
result2= mat1 @ mat2.T
result2
```

mat1 @ mat2 (when shapes are compatible):

Example: mat1 (2×3) @ mat2 (3×2)
┌─────────────────────────────────────────────────────┐
│                                                     │
│  mat1 = [[1, 2, 3],      mat2 = [[7, 8],          │
│          [4, 5, 6]]               [9, 10],         │
│                                   [11, 12]]        │
│                                                     │
│  Step-by-step multiplication:                      │
│                                                     │
│  Result[0,0] = [1,2,3] · [7,9,11] = 1×7+2×9+3×11 │
│              = 7 + 18 + 33 = 58                    │
│                                                     │
│  Result[0,1] = [1,2,3] · [8,10,12] = 1×8+2×10+3×12│
│              = 8 + 20 + 36 = 64                    │
│                                                     │
│  Result[1,0] = [4,5,6] · [7,9,11] = 4×7+5×9+6×11 │
│              = 28 + 45 + 66 = 139                  │
│                                                     │
│  Result[1,1] = [4,5,6] · [8,10,12] = 4×8+5×10+6×12│
│              = 32 + 50 + 72 = 154                  │
│                                                     │
│  Final Result: [[58, 64],                          │
│                 [139, 154]]                        │
└─────────────────────────────────────────────────────┘
