```python
import pandas as pd

# Base Series
names = pd.Series(["alice", "BOB", "Charlie"])

print("1. Uppercase:")
print(names.str.upper())

print("\n2. String Length:")
print(names.str.len())

print("\n3. Contains 'li':")
print(names.str.contains("li"))

print("\n4. Split by 'a':")
print(names.str.split("a"))
```
