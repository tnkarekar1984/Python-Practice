# To truly understand all three operations, we'll use a Series that includes a missing value (NaN). This makes the isnull() and fillna() operations meaningful and visual.

```python
import pandas as pd
import numpy as np

# Base Series with a Missing Value (NaN)
s = pd.Series([100, 200, np.nan, 400], index=["a", "b", "c", "d"])

print("Original Series:")
print(s)

print("\n1. Boolean Filtering (s > 200):")  # here it will print the value greater than 200
print(s[s > 200])

print("\n2. Checking Missing Values (s.isnull()):")  # here it will print the missing value
print(s.isnull())

print("\n3. Filling Missing Values (s.fillna(0)):")  # here it will fill the missing value with 0
print(s.fillna(0))
```
