```python
import pandas as pd

# Creating our sample Series
s = pd.Series([10, 20, 30, 40, 50], index=["a", "b", "c", "d", "e"])

print("Original Series:")
print(s)

# Let's explore ALL indexing methods
print("\n1. s['c']           →", s["c"])  # Label access
print("2. s[1:4]           →\n", s[1:4])  # Position slicing
print("3. s[['a', 'd']]    →\n", s[["a", "d"]])  # Multiple labels
print("4. s.loc['b']       →", s.loc["b"])  # Explicit label
print("5. s.iloc[3]        →", s.iloc[3])  # Explicit position
print("6. s.loc['b':'d']   →\n", s.loc["b":"d"])  # Label slicing
print("7. s.iloc[1:4]      →\n", s.iloc[1:4])  # Position slicing
```

# Critical Differences You Must Know

#1. Slicing Behavior: LOC vs ILOC




```python
## Position slicing (EXCLUDES end)
s[0:3]  # Gets positions 0, 1, 2 (NOT 3)
s.iloc[0:3]  # Same: positions 0, 1, 2
```

```python
## Label slicing (INCLUDES end)
s.loc["a":"c"]  # Gets 'a', 'b', 'c' (INCLUDES 'c')
```

# Single vs Multiple Bracket Magic

```python
s["b"]  # Single → returns value (int): 20
```

```python
s[["b"]]  # Double → returns Series: b    20
```

# 3. What Gets Returned?

```python
# Single element = VALUE (not Series)
type(s["a"])  # <class 'int'>
```

```python
type(s.loc["b"])  # <class 'int'>
```

```python
type(s.iloc[0])  # <class 'int'>

```

```python
# Multiple elements = SERIES
type(s[["a", "c"]])  # <class 'pandas.core.series.Series'>
```

```python
type(s[1:3])  # <class 'pandas.core.series.Series'>
```

# Boolean Indexing

```python

# Find all values > 25
s[s > 25]
```

# Negative Indexing

```python
# Access from the end
s.iloc[-1]  # Last element: 50
s.iloc[-2:]  # Last 2 elements: d=40, e=50
```

# 🧪 Practice Examples

```python
import pandas as pd

# Create Series
grades = pd.Series([85, 92, 78, 95, 88], index=["Alice", "Bob", "Charlie", "Diana", "Eve"])
```

```python
print(grades)  # 0    85
               # 1    92
```

```python
print(grades["Bob"])  # 92
```

```python
print(grades[["Bob", "Charlie"]])  # Bob    92
                                # Charlie   
```

```python
print(grades[grades > 90])  # Bob    92
                        # Diana   9
```

```python
print(grades[grades > 90].mean())  # 92.5
```
