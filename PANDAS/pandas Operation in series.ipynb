# Pandas Series Operations: A Complete Visual Guide

## What are Series Operations?

Pandas Series support vectorized operations (applying math to all elements at once without loops), aggregations (collapsing data to a single value), transformations (changing values), and method chaining (piping operations sequentially). Think of it as having a supercharged calculator for entire columns of data.


```python
import pandas as pd

# Base Series
s = pd.Series([1, 2, 3, 4])
```

# 1. Arithmetic


```python
print(s + 10)  # Add 10 to each element
print(s * 2)  # Element wise multiplication
print(s - 1)  # Subtract 1 from each element
print(s / 2)  # Element wise division
```

# 2. Aggregations


```python
print(s.mean())  # Average
print(s.max())  # Maximum
print(
    s.std()
)  # Standard deviation it has been calculated by the formula (n-1) * (sum of squares - (sum of elements)2/n)
```

# 3. Method Chaining (Pipeline)


```python
import pandas as pd

# Base Series
s = pd.Series([15, 25, 45, 65])

print("s as a pandas Series:")
print(s)
```

```python
result = (
    s.add(1).mul(10).clip(0, 420).add_suffix(" Result: ").add_prefix("Result: ")
)  # here we have used add, mul and clip method chaining what it does is it first add 1 to each element then multiply it by 10 and then clip the values between 0 and 25
print(result)
```

# 4. Removing Duplicates

```python
dup = pd.Series([1, 2, 2, 3])
print(dup.drop_duplicates())
```

# 5. Apply (Function) & Map (Dictionary)

```python
print(s.apply(lambda x: x**2))
print(s.map({1: "A", 2: "B", 3: "C"}))
```

```python
print(s.map({1: "A", 2: "B", 3: "C", 4: "D"}))
```

```python
s2 = pd.Series([1, 2, 3, 4])
print(s2.map({1: "A", 2: "B", 3: "C"}))
# Output: 0 A, 1 B, 2 C, 3 NaN  ← 4 becomes NaN!
```
